# Introduction to JetBot Nano

In this notebook, we will learn how to:

1. Access the JetBot camera.
2. Display the live camera feed inside Jupyter.
3. Control the JetBot motors.
4. Make the robot turn and move forward/backward.
5. Safely stop the camera when finished.


## 1. Import the required libraries

We first import the Python libraries needed for the camera and Jupyter widgets.

- `traitlets` lets us connect the camera output to an image widget.
- `ipywidgets` provides interactive widgets inside Jupyter.
- `display` displays widgets in the notebook.
- `Camera` gives us access to the JetBot camera.
- `bgr8_to_jpeg` converts camera images into JPEG format for display.


```python
import traitlets
import ipywidgets.widgets as widgets

from IPython.display import display
from jetbot import Camera, bgr8_to_jpeg
```

## 2. Start the camera

We create a camera object with a resolution of 224 × 224 pixels.

`Camera.instance()` creates or returns the JetBot camera instance.


```python
camera = Camera.instance(
    width=224,
    height=224
)
```

## 3. Create an image widget

The camera produces image data, but we need a Jupyter widget to display it in the notebook.

The widget below will display JPEG images at 224 × 224 pixels.


```python
image = widgets.Image(
    format='jpeg',
    width=224,
    height=224
)
```

## 4. Connect the camera to the image widget

The camera continuously produces images.

We use `traitlets.dlink()` to connect the camera output to the image widget.

The function `bgr8_to_jpeg` converts each camera frame into JPEG format before displaying it.


```python
camera_link = traitlets.dlink(
    (camera, 'value'),
    (image, 'value'),
    transform=bgr8_to_jpeg
)
```

## 5. Display the live camera feed

Running the following cell should display the JetBot camera feed directly inside the notebook.


```python
display(image)
```

# Controlling the JetBot Motors

Next, we will control the JetBot's motors.

Before running the following cells:

- Place the JetBot on the floor or another safe surface.
- Make sure there are no obstacles nearby.
- Keep the movement speed low while experimenting.


## 6. Import the Robot class

The `Robot` class provides simple commands for controlling the JetBot motors.

We also import Python's `time` module so that we can control how long each movement lasts.


```python
from jetbot import Robot
import time
```

## 7. Create the robot object

We create a `Robot` object that we will use to send commands to the motors.


```python
robot = Robot()
```

## 8. Turn left

The following code makes the JetBot turn left at a speed of `0.2`.

It moves for one second and then stops.


```python
robot.left(speed=0.2)

time.sleep(1)

robot.left(speed=0)
```

### Try it

Experiment with the speed and duration.

For example:

- What happens if the speed is changed from `0.2` to `0.1`?
- What happens if the robot turns for `0.5` seconds instead of `1` second?


## 9. Turn right

Now we perform the same experiment in the opposite direction.


```python
robot.right(speed=0.2)

time.sleep(1)

robot.right(speed=0)
```

## 10. Move forward

`set_motors(left_speed, right_speed)` lets us directly control both motors.

Setting both motors to the same positive speed makes the JetBot move forward.


```python
robot.set_motors(0.2, 0.2)

time.sleep(1)

robot.set_motors(0, 0)
```

### Understanding `set_motors`

The command

`robot.set_motors(left_speed, right_speed)`

controls the two motors independently.

For example:

- `(0.2, 0.2)` → both motors move forward.
- `(0, 0)` → both motors stop.
- `(-0.2, -0.2)` → both motors move backward.

Later, we can use different speeds for the left and right motors to steer the robot.


## 11. Move backward

Negative motor speeds make the wheels rotate in the opposite direction.

The following cell moves the JetBot backward for one second.


```python
robot.set_motors(-0.2, -0.2)

time.sleep(1)

robot.set_motors(0, 0)
```

## 12. Stop the robot

It is good practice to explicitly stop the motors after an experiment.


```python
robot.set_motors(0, 0)
```

# Shutting Down the Camera

When we are finished using the camera, we should stop it so that the camera resource is released.


```python
camera.stop()
```
